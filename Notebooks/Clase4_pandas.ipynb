## Repaso clases en python

En Python, una clase (**class**) es una estructura que permite agrupar datos (atributos) y funciones (métodos) dentro de un mismo objeto.
Una clase funciona como una plantilla para crear objetos que comparten ciertas características y comportamientos.

```
class clase:
  def __init__(self):
      #Aqui puedes definirlas variables de tu clase
  def fun(self):
    return n
```





````python
class Estrella:
    def __init__(self, nombre, masa, luminosidad):
        self.nombre = nombre
        self.masa = masa
        self.luminosidad = luminosidad

    def mostrar_info(self):
        print(f"Estrella: {self.nombre}")
        print(f"Masa: {self.masa}")
        print(f"Luminosidad: {self.luminosidad}")
````

class Estrella: define una nueva clase llamada Estrella.

*   __init__() es el método inicializador que se ejecuta cuando creamos un objeto.
*   self representa la instancia particular del objeto.
* self.nombre, self.masa y self.luminosidad son atributos.
* mostrar_info() es un método que permite mostrar información de la estrella.

















````python
estrella1 = Estrella(nombre="Sol", masa =1.0, luminosidad = 1.0)
estrella2 = Estrella(nombre="Sirio A", masa = 2.06, luminosidad = 25.4)

estrella1.mostrar_info()
estrella2.mostrar_info()
````

````python
print(estrella1.masa)  # 1.0
print(estrella2.masa)  # 2.06
````

````python
class Estrella:
    def __init__(self, nombre, luminosidad, distancia):
        self.nombre = nombre
        self.luminosidad = luminosidad
        self.distancia = distancia

    def calcular_flujo(self):
        F = self.luminosidad / (4 * (3.14)* self.distancia**2)
        return F

estrella = Estrella(nombre = "Sol", luminosidad=3.828e26, distancia=1.496e11)

print(estrella.calcular_flujo())
````
