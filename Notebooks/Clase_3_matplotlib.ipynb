#### Comó se escribir funciones en python

```python
#De ahora al futuro cuando escriban funciones en las tareas este es el formato que debe seguir, explicando que hace la funcion,
#luego definiendo cada una de las variables y el tipo que puede ser utilizado y el tipo que retornara
def multiplicar(numero, multiplicador=2):
    """
    Multiplica un número o arreglo por un multiplicador.

    Parameters
    ----------
    numero : int, float or numpy.ndarray
        Número o arreglo de números que se desea multiplicar.
    multiplicador : int or float, optional
        Valor por el cual se multiplicará `numero`.
        Por defecto es 2.

    Returns
    -------
    int, float or numpy.ndarray
        Resultado de la multiplicación.
    """
    resultado = numero * multiplicador
    return resultado
```

```python
plt.plot(x1,y1,label="distribucion 1")
plt.plot(x2,y2,label="distribucion 2")
plt.legend()
```

```python
plt.plot(x1,y1,label="distribucion 1",zorder=1,marker="*")
plt.plot(x2,y2,label="distribucion 2",zorder=0,marker = '.')
plt.xlabel("distribucion eje x")
plt.xlabel("distribucion eje y")
plt.legend()
```

```python
plt.plot(x1,y1,label="distribucion 1",zorder=1,marker="*",ls= "None",markersize=15)
plt.plot(x2,y2,label="distribucion 2",zorder=0,marker = '.',ls = "None",markersize=15)
plt.text(0, 0, 'escribi dentro de un plot', style='italic', bbox={'facecolor': 'white', 'alpha': 0.5, 'pad': 2})
#alpha mide lo difuso, pad el ancho del recuadro, y facecolor el color de la caja
plt.xlabel("distribucion eje x",fontsize =15)
plt.ylabel("distribucion eje y",fontsize =15)
plt.title('titulo del plot',fontsize =15)
plt.tick_params(
    axis="both",       # "x", "y", or "both"
    which="major",     # "major" or "minor"
    direction="out",    # "in", "out", or "inout"
    length=6,           #largo del tick
    width=1.5,          # ancho del tick
    top=False,
    right=False,
    labelsize=20 )
#plt.xlim(-2,2)
#plt.ylim(-2,2)
plt.legend(fontsize =15)
plt.show()
```

Tambien matplotlib tiene la opcion de hacer multiples figuras al mismo tiempo

```python
fig, axs = plt.subplots(2, 1, figsize=(10, 10), tight_layout=True)
axs[0].plot(x1, y1)
axs[1].plot(x2, y2)
ticks = np.arange(0., 8.1, 2.)

tickla = [f'{tick:1.2f}' for tick in ticks]
axs[1].xaxis.set_ticks(ticks)
axs[1].xaxis.set_ticklabels(tickla)
axs[1].set_xlim(axs[0].get_xlim())
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 10), tight_layout=True)
axs[0].plot(x1, y1)
axs[1].plot(x2, y2)
ticks = np.arange(0., 8.1, 2.)

tickla = [f'{tick:1.2f}' for tick in ticks]
axs[1].xaxis.set_ticks(ticks)
axs[1].xaxis.set_ticklabels(tickla)
axs[1].set_xlim(axs[0].get_xlim())
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 5), tight_layout=True)
axs[0].plot(x1,y1,label="distribucion 1",zorder=1,marker="*",ls= "None",markersize=15)
axs[1].plot(x2,y2,label="distribucion 2",zorder=0,marker = '.',ls = "None",markersize=15)
axs[0].text(1, 0, 'plot1',  transform= axs[0].transAxes,fontsize=20)  # cuando se transforma a las coordenadas de la figura van de [0,1]

axs[0].set_xlabel("distribucion eje x",fontsize =15)
axs[0].set_ylabel("distribucion eje y",fontsize =15)
axs[0].set_title('titulo del plot',fontsize =15)

axs[0].set_xlim(-2,2)
axs[0].set_ylim(-2,2)

axs[1].set_xlim(-2,2)
axs[1].set_ylim(-2,2)


axs[0].tick_params(
    axis="both",       # "x", "y", or "both"
    which="major",     # "major" or "minor"
    direction="out",    # "in", "out", or "inout"
    length=6,           #largo del tick
    width=1.5,          # ancho del tick
    top=False,
    right=False,
    labelsize=20 )

axs[1].tick_params(
    axis="both",
    which="major",
    direction="out",
    length=6,
    width=1.5,
    top=False,
    right=False,
    labelsize=20 )

axs[0].legend(fontsize =15)
axs[1].legend(fontsize =15)

plt.show()
```

## Entonces como se hacen figuras de imagenes (o 2D con alguna intensidad)

```python
#generados los datos
X, Y = np.meshgrid(np.linspace(-3, 3, 16), np.linspace(-3, 3, 16))
Z = (1 - X/2 + X**5 + Y**3) * np.exp(-X**2 - Y**2)

# figura
fig, ax = plt.subplots(figsize=(5, 5))

im = ax.imshow(Z, origin="lower", vmax=np.max(Z)) #Los arreglos NumPy arrays presentados con imshow(origin="lower"), row i=0 appears at the bottom, not the top.

# Agregamos una barra de color
cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Intensity", fontsize=15)
cbar.ax.tick_params(labelsize=15)

ax.tick_params(labelsize=20)

plt.show()
```

Cada posision (i,j) contiene un coordenada par (X[i,j], Y[i,J]).

```python
from itertools import product #para conseguir todas las posibles combinaciones
indices_posibles = np.arange(x.size)
combinations = list(product(indices_posibles, indices_posibles))
#i corresponde al eje Y, mientras j representa al eje X en este ejemplo
print("X[i,j]    Y[i,j]")
for i,j in combinations:
  print(f"X[{i},{j}] = {X[i,j]},Y[{i},{j}] = {Y[i,j]}")
```

```python
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

X, Y = np.meshgrid(x, y)

x0 = [-1.5, 1.5]
y0 = [-1.0, 1.0]


Z1 = np.exp(-((X - x0[0])**2 + (Y - y0[0])**2) / 2)
Z2 = np.exp(-((X - x0[1])**2 + (Y - y0[1])**2) / 2)

Z_individual = Z1 + Z2


Z = np.exp(-((X[:, :, None] - x0)**2 + (Y[:, :, None] - y0)**2) / 2)

Z_broadcast = np.sum(Z, axis=2)

fig, axs = plt.subplots(1, 2, figsize=(12, 5),sharex=True,sharey=True,constrained_layout=True)

vmax = max(np.max(Z_individual), np.max(Z_broadcast))

for ax, data, title in zip(axs,[Z_individual, Z_broadcast],["Individual Gaussians", "NumPy Broadcasting"]):
    im = ax.imshow(data,origin="lower",extent=[x.min(), x.max(), y.min(), y.max()],cmap="inferno", vmin=0, vmax=vmax)
    ax.set_xticks(np.arange(-3, 4, 1))
    ax.set_yticks(np.arange(-3, 4, 1))
    ax.set_xlabel("X", fontsize=15)
    ax.tick_params(labelsize=15)
    ax.set_title(title)

axs[0].set_ylabel("Y", fontsize=15)


cbar = fig.colorbar(im,ax=axs,location="right",shrink=1.,pad=0.02)

cbar.set_label("Intensity", fontsize=15)
cbar.ax.tick_params(labelsize=15)

plt.show()
```
