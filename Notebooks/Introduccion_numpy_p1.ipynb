## Interludio
  ¿Como escribir funciones?


```python
#De ahora al futuro cuando escriban funciones en las tareas este es el formato que debe seguir, explicando que hace la funcion,
#luego definiendo cada una de las variables y el tipo que puede ser utilizado y el tipo que retornara
def multiplicar(numero, multiplicador=2):
    """
    Multiplica un número o arreglo por un multiplicador.

    Parameters
    ----------
    numero : int, float or numpy.ndarray
        Número o arreglo de números que se desea multiplicar.
    multiplicador : int or float, optional
        Valor por el cual se multiplicará `numero`.
        Por defecto es 2.

    Returns
    -------
    int, float or numpy.ndarray
        Resultado de la multiplicación.
    """
    resultado = numero * multiplicador
    return resultado
```

```python
multiplicar(np.array([2,2,3]))
```

## Condicionales en numpy

```python
arreglo_numpy_2D = np.array([[1,2,3],[3,2,1]])
arreglo_numpy_2D
```

```python
mascara = arreglo_numpy_2D<=1
mascara
```

```python
arreglo_numpy_2D[mascara] = 100
```

```python
arreglo_numpy_2D
```

```python
np.where(arreglo_numpy_2D<=2)
```

```python
# Entonces la funcion np.where no solo puede encontrar los indices tambien nos puede ayudar a moficar valores del arreglo mismo
#condicion, si la condicion se cumple el valor sera reemplazado por, en caso de que no se cumple el valor que se quedara sera
np.where(arreglo_numpy_2D<=2,0,arreglo_numpy_2D)
```

```python
np.where(arreglo_numpy_2D<=2,0,1)
```

## Combinando arreglos

```python
arreglo1 = np.ones(4)
arreglo0 = np.zeros(4)

arreglo01 = np.stack([arreglo1,arreglo0],axis=0)
arreglo01
```

```python
arreglo01 = np.stack([arreglo1,arreglo0],axis=1)
arreglo01
```

```python
arreglo01 = np.concat([arreglo1,arreglo0],axis=0)
arreglo01
```

```python
arreglo1 = np.ones((4,2))
arreglo0 = np.zeros((4,2))
```

```python
arreglo01 = np.concat([arreglo1,arreglo0],axis=1)
arreglo01
```

```python
# si quieres crear arreglos de zeros o unos con el mismo tamaño que otro

unos = np.ones_like(arreglo01)
ceros = np.zeros_like(arreglo01)

print(arreglo01.shape,unos.shape,ceros.shape)
```

## Seudo generador de numeros aleatorios

```python
#Numpy tambien nos permite crear arreglos entre dos limites con una cierta cantidad de elementos
arreglo_numeros=np.linspace(0,10,100)
np.shape(arreglo_numeros)
```

```python
rng = np.random.default_rng(seed=0)
```

```python
numeros = rng.uniform(low=1,high=10,size=(1000,))
plt.hist(numeros)
```

```python
numeros_gaussianos = rng.normal(loc=0,scale=2,size=(100,))
plt.hist(numeros_gaussianos)
```

```python
np.mean(numeros_gaussianos)
```

```python
np.median(numeros_gaussianos)
```

```python
np.std(numeros_gaussianos)
```

```python
np.argmax(numeros_gaussianos)
```

```python
np.argmin(numeros_gaussianos)
```

```python
#de menor a mayor
np.argsort(numeros_gaussianos,axis=0)
```
